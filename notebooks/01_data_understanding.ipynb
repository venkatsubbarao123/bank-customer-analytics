# 01 - Data Understanding
Profile the original CSV without modifying it.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW_PATH = ROOT / 'data' / 'raw' / 'bank_transactions.csv'
df = pd.read_csv(RAW_PATH, low_memory=False)
report = {
    'Rows': len(df),
    'Columns': len(df.columns),
    'Unique Customers': df['CustomerID'].nunique(),
    'Unique Transactions': df['TransactionID'].nunique(),
    'Date Range': f"{df['TransactionDate'].min()} to {df['TransactionDate'].max()}",
    'Average Transaction INR': df['TransactionAmount (INR)'].mean(),
    'Maximum Transaction INR': df['TransactionAmount (INR)'].max(),
    'Duplicate Rows': int(df.duplicated().sum()),
}
pd.Series(report)
```

```python
missing = df.isna().sum().sort_values(ascending=False).rename('missing_values').to_frame()
missing['missing_percent'] = (missing['missing_values'] / len(df) * 100).round(2)
display(missing)
display(df.dtypes.rename('data_type').to_frame())
print('Duplicate transaction IDs:', df['TransactionID'].duplicated().sum())
```

```python
display(df.head())
print(f'Raw source preserved at: {RAW_PATH}')
```
